```python
import pandas as pd
import numpy as np
import os

DATA_DIR = "cic_ids_2018"
SAMPLE_PER_CLASS = 40000

# 16 standard behavioral network features
CORE_FEATURES = [
    'Dst Port', 'Flow Duration', 'Tot Fwd Pkts', 'Tot Bwd Pkts',
    'TotLen Fwd Pkts', 'TotLen Bwd Pkts', 'Fwd Pkt Len Mean',
    'Bwd Pkt Len Mean', 'Flow Byts/s', 'Flow Pkts/s',
    'Flow IAT Mean', 'Flow IAT Std', 'SYN Flag Cnt',
    'ACK Flag Cnt', 'RST Flag Cnt', 'Init Fwd Win Byts'
]

def load_and_sample(filename, label_filter, stage_id, sample_count):
    path = os.path.join(DATA_DIR, filename)
    print(f"Loading Stage {stage_id} from {filename}...")
    chunks = []
    total_found = 0
    for chunk in pd.read_csv(path, chunksize=100000, low_memory=False):
        chunk.columns = chunk.columns.str.strip()

        if label_filter == 'Benign':
            matched = chunk[chunk['Label'].astype(str).str.lower() == 'benign']
        elif label_filter == 'BruteForce':
            matched = chunk[chunk['Label'].astype(str).str.contains('BruteForce|Bruteforce', case=False, na=False)]
        elif label_filter == 'Bot':
            matched = chunk[chunk['Label'].astype(str).str.contains('Bot', case=False, na=False)]
        elif label_filter == 'DDOS':
            matched = chunk[chunk['Label'].astype(str).str.contains('DDOS', case=False, na=False)]

        if not matched.empty:
            chunks.append(matched)
            total_found += len(matched)
            if total_found >= sample_count * 2:
                break

    df = pd.concat(chunks, ignore_index=True)
    sampled = df.sample(n=min(sample_count, len(df)), random_state=42).copy()
    sampled['Stage'] = stage_id
    return sampled

# --- Extracting All 4 Stages Correctly ---

# Stage 0: Benign (from Thursday 15-02, which has 147k benign flows)
df_0 = load_and_sample("Thursday-15-02-2018_TrafficForML_CICFlowMeter.csv", "Benign", 0, SAMPLE_PER_CLASS)

# Stage 1: Initial Access / Brute Force (from Wednesday 14-02, which has 199k brute force flows)
df_1 = load_and_sample("Wednesday-14-02-2018_TrafficForML_CICFlowMeter.csv", "BruteForce", 1, SAMPLE_PER_CLASS)

# Stage 2: Botnet / C2 (from Friday 02-03, which has 162k bot flows)
df_2 = load_and_sample("Friday-02-03-2018_TrafficForML_CICFlowMeter.csv", "Bot", 2, SAMPLE_PER_CLASS)

# Stage 3: Volumetric DDoS (from Wednesday 21-02, which has 196k DDoS flows)
df_3 = load_and_sample("Wednesday-21-02-2018_TrafficForML_CICFlowMeter.csv", "DDOS", 3, SAMPLE_PER_CLASS)

# --- Merge and Clean ---
full_df = pd.concat([df_0, df_1, df_2, df_3], ignore_index=True)
full_df.columns = full_df.columns.str.strip()

# Normalize column names
rename_map = {
    'Destination Port': 'Dst Port', 'Total Fwd Packets': 'Tot Fwd Pkts',
    'Total Backward Packets': 'Tot Bwd Pkts', 'Total Length of Fwd Packets': 'TotLen Fwd Pkts',
    'Total Length of Bwd Packets': 'TotLen Bwd Pkts', 'Fwd Packet Length Mean': 'Fwd Pkt Len Mean',
    'Bwd Packet Length Mean': 'Bwd Pkt Len Mean', 'Flow Bytes/s': 'Flow Byts/s',
    'Flow Packets/s': 'Flow Pkts/s', 'SYN Flag Count': 'SYN Flag Cnt',
    'ACK Flag Count': 'ACK Flag Cnt', 'RST Flag Count': 'RST Flag Cnt',
    'Init_Win_bytes_forward': 'Init Fwd Win Byts'
}
full_df.rename(columns=rename_map, inplace=True)

# Keep standard features + Stage
matched_features = [col for col in CORE_FEATURES if col in full_df.columns]
clean_df = full_df[matched_features + ['Stage']].copy()

# Remove infinities and NaNs
clean_df[matched_features] = clean_df[matched_features].apply(pd.to_numeric, errors='coerce')
clean_df.replace([np.inf, -np.inf], np.nan, inplace=True)
clean_df.dropna(inplace=True)

# Save the balanced file
clean_df.to_csv("balanced_multistage_2018.csv", index=False)

print("\n--- Success! Balanced Dataset Summary ---")
print(clean_df['Stage'].value_counts())
print(f"Total Rows Saved: {len(clean_df)}")
```

```python
import numpy as np
import pandas as pd
import joblib
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# 1. 15 Behavioral Dynamic Features (NO Destination Port shortcut)
DYNAMIC_FEATURES = [
    'Flow Duration', 'Tot Fwd Pkts', 'Tot Bwd Pkts',
    'TotLen Fwd Pkts', 'TotLen Bwd Pkts', 'Fwd Pkt Len Mean',
    'Bwd Pkt Len Mean', 'Flow Byts/s', 'Flow Pkts/s',
    'Flow IAT Mean', 'Flow IAT Std', 'SYN Flag Cnt',
    'ACK Flag Cnt', 'RST Flag Cnt', 'Init Fwd Win Byts'
]

df = pd.read_csv("balanced_multistage_2018.csv")

# 2. Per-Class Chronological Partitioning (Guarantees all 4 classes in Train, Val, Test)
train_dfs, val_dfs, test_dfs = [], [], []

for stage in [0, 1, 2, 3]:
    stage_df = df[df['Stage'] == stage].copy()
    n = len(stage_df)
    n_train = int(n * 0.70)
    n_val   = int(n * 0.85)

    train_dfs.append(stage_df.iloc[:n_train])
    val_dfs.append(stage_df.iloc[n_train:n_val])
    test_dfs.append(stage_df.iloc[n_val:])

train_df = pd.concat(train_dfs, ignore_index=True)
val_df   = pd.concat(val_dfs, ignore_index=True)
test_df  = pd.concat(test_dfs, ignore_index=True)

print("Class counts in TEST set (All 4 present):")
print(test_df['Stage'].value_counts())

# 3. Fit Scaler ONLY on Training data
scaler = StandardScaler()
X_train_raw = train_df[DYNAMIC_FEATURES].values
y_train_raw = train_df['Stage'].values

X_val_raw   = val_df[DYNAMIC_FEATURES].values
y_val_raw   = val_df['Stage'].values

X_test_raw  = test_df[DYNAMIC_FEATURES].values
y_test_raw  = test_df['Stage'].values

X_train_scaled = scaler.fit_transform(X_train_raw)
X_val_scaled   = scaler.transform(X_val_raw)
X_test_scaled  = scaler.transform(X_test_raw)

joblib.dump(scaler, "scaler_2018.pkl")

# 4. Generate Sliding Sequences independently (Zero overlap leakage)
def create_sequences(data, labels, window_size=20, stride=5):
    X_seq, y_next_state, y_class = [], [], []
    for i in range(0, len(data) - window_size - 1, stride):
        X_seq.append(data[i : i + window_size])
        y_next_state.append(data[i + window_size])
        y_class.append(labels[i + window_size])
    return (
        np.array(X_seq, dtype=np.float32),
        np.array(y_next_state, dtype=np.float32),
        np.array(y_class, dtype=np.int64)
    )

X_train, s_train, c_train = create_sequences(X_train_scaled, y_train_raw)
X_val,   s_val,   c_val   = create_sequences(X_val_scaled,   y_val_raw)
X_test,  s_test,  c_test  = create_sequences(X_test_scaled,  y_test_raw)

print(f"\nSequences created -> Train: {len(X_train)}, Val: {len(X_val)}, Test: {len(X_test)}")

# 5. Model Definition
class NetworkWorldModel(nn.Module):
    def __init__(self, input_dim=15, hidden_dim=64, num_classes=4, dropout=0.2):
        super(NetworkWorldModel, self).__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers=2, batch_first=True, dropout=dropout)

        self.dynamics_head = nn.Sequential(
            nn.Linear(hidden_dim, 32),
            nn.ReLU(),
            nn.Linear(32, input_dim)
        )

        self.classifier_head = nn.Sequential(
            nn.Linear(hidden_dim, 32),
            nn.ReLU(),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        lstm_out, _ = self.lstm(x)
        h_t = lstm_out[:, -1, :]
        next_state = self.dynamics_head(h_t)
        logits = self.classifier_head(h_t)
        return next_state, logits

train_loader = DataLoader(TensorDataset(torch.tensor(X_train), torch.tensor(s_train), torch.tensor(c_train)), batch_size=256, shuffle=True)
val_loader   = DataLoader(TensorDataset(torch.tensor(X_val),   torch.tensor(s_val),   torch.tensor(c_val)),   batch_size=256, shuffle=False)
test_loader  = DataLoader(TensorDataset(torch.tensor(X_test),  torch.tensor(s_test),  torch.tensor(c_test)),  batch_size=256, shuffle=False)

model = NetworkWorldModel(input_dim=len(DYNAMIC_FEATURES), hidden_dim=64, num_classes=4).to(device)
criterion_dynamics = nn.MSELoss()
criterion_class = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("\n--- Training Dual-Head World Model (8 Epochs) ---")
for epoch in range(8):
    model.train()
    tot_loss, tot_dyn, tot_cls = 0, 0, 0
    for bx, bs, bc in train_loader:
        bx, bs, bc = bx.to(device), bs.to(device), bc.to(device)
        optimizer.zero_grad()
        pred_state, logits = model(bx)

        l_dyn = criterion_dynamics(pred_state, bs)
        l_cls = criterion_class(logits, bc)
        loss = l_dyn + l_cls

        loss.backward()
        optimizer.step()

        tot_loss += loss.item()
        tot_dyn += l_dyn.item()
        tot_cls += l_cls.item()

    print(f"Epoch [{epoch+1}/8] | Loss: {tot_loss/len(train_loader):.4f} | Dyn MSE: {tot_dyn/len(train_loader):.4f} | Cls NLL: {tot_cls/len(train_loader):.4f}")

# 6. Temperature Calibration (T optimization)
model.eval()
val_logits_list, val_labels_list = [], []
with torch.no_grad():
    for bx, _, bc in val_loader:
        _, logits = model(bx.to(device))
        val_logits_list.append(logits)
        val_labels_list.append(bc.to(device))

val_logits = torch.cat(val_logits_list)
val_labels = torch.cat(val_labels_list)

temperature = nn.Parameter(torch.ones(1, device=device) * 1.5)
optimizer_temp = optim.LBFGS([temperature], lr=0.01, max_iter=50)

def eval_temp():
    optimizer_temp.zero_grad()
    loss = criterion_class(val_logits / temperature, val_labels)
    loss.backward()
    return loss

optimizer_temp.step(eval_temp)
optimal_T = float(temperature.item())
print(f"\n✅ Optimal Calibration Temperature T = {optimal_T:.4f}")

# 7. Complete 4-Stage Test Evaluation
test_preds, test_true = [], []
with torch.no_grad():
    for bx, _, bc in test_loader:
        _, logits = model(bx.to(device))
        calibrated_probs = torch.softmax(logits / optimal_T, dim=1)
        preds = torch.argmax(calibrated_probs, dim=1)
        test_preds.extend(preds.cpu().numpy())
        test_true.extend(bc.numpy())

target_names = [
    "Stage 0 (Nominal)",
    "Stage 1 (BruteForce/Recon)",
    "Stage 2 (Botnet/C2)",
    "Stage 3 (DDoS/Impact)"
]
print("\n=== COMPLETE 4-STAGE OUT-OF-SAMPLE TEST EVALUATION ===")
print(classification_report(test_true, test_preds, target_names=target_names, digits=4))
print("Confusion Matrix (4x4):")
print(confusion_matrix(test_true, test_preds))

# 8. Save Artifacts for Streamlit App
torch.save({'model_state': model.state_dict(), 'optimal_T': optimal_T}, "network_world_model_2018.pth")
np.savez_compressed("dashboard_test_stream_2018.npz", X_test=X_test[:2000], y_test=c_test[:2000])
print("\nAll artifacts saved successfully!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.metrics import classification_report, confusion_matrix

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# 1. Dual-Head Network World Model Definition
class NetworkWorldModel(nn.Module):
    def __init__(self, input_dim=16, hidden_dim=64, num_classes=4, dropout=0.2):
        super(NetworkWorldModel, self).__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers=2, batch_first=True, dropout=dropout)

        # Dynamics Head: Predicts continuous 16 features of next state S_{t+1}
        self.dynamics_head = nn.Sequential(
            nn.Linear(hidden_dim, 32),
            nn.ReLU(),
            nn.Linear(32, input_dim)
        )

        # Tactical Head: Predicts discrete MITRE attack stage (0, 1, 2, 3)
        self.classifier_head = nn.Sequential(
            nn.Linear(hidden_dim, 32),
            nn.ReLU(),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        lstm_out, _ = self.lstm(x)
        h_t = lstm_out[:, -1, :]  # Final hidden state across the 20 steps
        next_state_pred = self.dynamics_head(h_t)
        class_logits = self.classifier_head(h_t)
        return next_state_pred, class_logits

# 2. PyTorch DataLoaders
train_ds = TensorDataset(torch.tensor(X_train), torch.tensor(s_train), torch.tensor(c_train))
val_ds   = TensorDataset(torch.tensor(X_val),   torch.tensor(s_val),   torch.tensor(c_val))
test_ds  = TensorDataset(torch.tensor(X_test),  torch.tensor(s_test),  torch.tensor(c_test))

train_loader = DataLoader(train_ds, batch_size=256, shuffle=True)
val_loader   = DataLoader(val_ds,   batch_size=256, shuffle=False)
test_loader  = DataLoader(test_ds,  batch_size=256, shuffle=False)

# 3. Model Training Loop
model = NetworkWorldModel(input_dim=16, hidden_dim=64, num_classes=4).to(device)
criterion_dynamics = nn.MSELoss()
criterion_class = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("\n--- Training Dual-Head World Model (8 Epochs) ---")
epochs = 8
for epoch in range(epochs):
    model.train()
    total_loss, total_dyn_loss, total_cls_loss = 0, 0, 0
    for bx, bs, bc in train_loader:
        bx, bs, bc = bx.to(device), bs.to(device), bc.to(device)
        optimizer.zero_grad()

        pred_state, logits = model(bx)
        l_dyn = criterion_dynamics(pred_state, bs)
        l_cls = criterion_class(logits, bc)
        loss = l_dyn + l_cls

        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        total_dyn_loss += l_dyn.item()
        total_cls_loss += l_cls.item()

    print(f"Epoch [{epoch+1}/{epochs}] | Loss: {total_loss/len(train_loader):.4f} | Dynamics MSE: {total_dyn_loss/len(train_loader):.4f} | Class NLL: {total_cls_loss/len(train_loader):.4f}")

# 4. Temperature Calibration (T optimization on Validation Set)
print("\n--- Optimizing Temperature Calibration Parameter (T) ---")
model.eval()
val_logits_list, val_labels_list = [], []
with torch.no_grad():
    for bx, _, bc in val_loader:
        _, logits = model(bx.to(device))
        val_logits_list.append(logits)
        val_labels_list.append(bc.to(device))

val_logits = torch.cat(val_logits_list)
val_labels = torch.cat(val_labels_list)

# Define learned scalar T
temperature = nn.Parameter(torch.ones(1, device=device) * 1.5)
optimizer_temp = optim.LBFGS([temperature], lr=0.01, max_iter=50)

def eval_temp():
    optimizer_temp.zero_grad()
    loss = criterion_class(val_logits / temperature, val_labels)
    loss.backward()
    return loss

optimizer_temp.step(eval_temp)
optimal_T = float(temperature.item())
print(f"✅ Learned Optimal Temperature T = {optimal_T:.4f}")

# 5. Out-of-Sample Test Evaluation on All 4 Stages
test_preds, test_true = [], []
with torch.no_grad():
    for bx, _, bc in test_loader:
        _, logits = model(bx.to(device))
        # Calibrate logits with temperature T
        calibrated_probs = torch.softmax(logits / optimal_T, dim=1)
        preds = torch.argmax(calibrated_probs, dim=1)
        test_preds.extend(preds.cpu().numpy())
        test_true.extend(bc.numpy())

target_names = [
    "Stage 0 (Nominal)",
    "Stage 1 (BruteForce/Recon)",
    "Stage 2 (Botnet/C2)",
    "Stage 3 (DDoS/Impact)"
]
print("\n=== COMPLETE 4-STAGE OUT-OF-SAMPLE TEST EVALUATION ===")
print(classification_report(test_true, test_preds, target_names=target_names, digits=4))
print("Confusion Matrix (4x4):")
print(confusion_matrix(test_true, test_preds))

# 6. Save Model Artifacts
torch.save({
    'model_state': model.state_dict(),
    'optimal_T': optimal_T
}, "network_world_model_2018.pth")
print("\nSaved network_world_model_2018.pth successfully!")
```
